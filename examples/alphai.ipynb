## Call the tools directly

```python
from llama_index.tools.alphai import AlphaAIToolSpec

spec = AlphaAIToolSpec()

docs = spec.search_news(symbol="NVDA", min_relevance=7, max_results=5)
for doc in docs:
    print(f"[{doc.metadata['relevance_score']}/10] {doc.metadata['title']}")
```

```python
events = spec.get_insider_news(min_relevance=7, max_results=5)
for event in events:
    insider = event.metadata["insider"]
    print(
        f"{event.metadata['tickers']}: {insider['insider_name']} "
        f"({insider['insider_title']}) {insider['side']} "
        f"${insider['total_value_usd']}"
    )
```

```python
spec.get_ticker_sentiment("NVDA")
```

## Use in an agent

```python
from llama_index.core.agent.workflow import FunctionAgent
from llama_index.llms.openai import OpenAI

agent = FunctionAgent(
    tools=AlphaAIToolSpec().to_tool_list(),
    llm=OpenAI(model="gpt-5"),
)

response = await agent.run(
    "Are NVDA insiders net buying or selling this month, "
    "and does news sentiment agree with what they're doing?"
)
print(response)
```
